```python
import pandas as pd
dfriso=pd.read_csv("outputs/per_protein_diamond_data.tsv",sep="\t",index_col=0)
```

```python
htgs= set(dfriso[dfriso.tree_HTG].accession)
htg_genes=set(dfriso[dfriso.tree_HTG].gene)
```

```python
gff=pd.read_csv('GCA_965233685.1_liftoff/annotation_polished',sep='\t',header=None,skiprows=3)
```

```python
cds=gff[gff[2]=='CDS']
prot_transcript={x.split(";")[0].split("cds-")[1]:x.split(";")[1].split("rna-")[1] for x in cds[8]}
transcript_prot={x.split(";")[1].split("rna-")[1]:x.split(";")[0].split("cds-")[1] for x in cds[8]}
```

```python
len([x for x in htgs if x not  in prot_transcript])
```

```python
comparison_df=pd.DataFrame()
```

```python
mRNA=gff[gff[2]=='mRNA']
mRNA['transcript']=[x.split(";")[0].split("-")[1] for x in mRNA[8]]
mRNA['gene']=[x.split("gene-")[1].split(';')[0] for x in mRNA[8]]
mRNA=mRNA[mRNA.transcript.isin([prot_transcript[x] for x in htgs if x in prot_transcript])]
# mRNA['matches_ref_protein']=[x.split(";matches_ref_protein=")[1].split(";")[0] for x in mRNA[8]]
for index, row in mRNA.iterrows():
    transcript=row.transcript
    prot=transcript_prot[transcript]
    if 'valid_ORF=False' in row[8]:
        comparison_df.loc[prot,'GCA_965233685.1_liftoff']='partial_ORF'
    elif row[8].split(";matches_ref_protein=")[1].split(";")[0]=='True':
        comparison_df.loc[prot,'GCA_965233685.1_liftoff']='full'
    else:
        comparison_df.loc[prot,'GCA_965233685.1_liftoff']='modified_ORF'
for y in [x for x in htgs if x not in prot_transcript]:
     comparison_df.loc[y,'GCA_965233685.1_liftoff']='no_transfer'
```

```python
comparison_df.to_csv("outputs/GCA_965233685.1_liftoff_comparison.tsv",sep="\t")
```
